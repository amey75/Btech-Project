## 5b. Evidence contradiction (all four models)
Replaces ONE evidence word with a conflicting one (man → woman, pool → ocean, red → blue). An independent NLI model
(DeBERTa-v3-large, the *oracle*) gives the label each edited pair should now have; the model is scored on whether it follows.

**Pilot first** (cell below, ~5 min): 300 pairs with random nt8 into a separate folder. Open
`spotcheck_contradiction.csv` from that folder, and each team member labels ~50 rows (is the edit natural? which label is right?).
Then run the four full models. Each full run re-uses that model's existing folder, so only the new edits are computed
(~15–25 min per model; the oracle cache in `btp_results/oracle_cache/` is shared, so later models are faster).

```
%cd /content/BTech-Project
# Pilot: 300 pairs, random nt8, separate folder
PILOT_DIR = f'{RESULTS}/pilot_contradiction_nt8'
!python faithfulness/run_experiment.py --model_path /content/drive/MyDrive/nt8 --n 300 --out_dir "$PILOT_DIR"
import pandas as pd
ed = pd.read_csv(f'{PILOT_DIR}/contradiction_edits.csv')
ed[['variant','op','target','substitute','edited_premise','edited_hypothesis','oracle_label','oracle_conf','model_pred_on_edit']].sample(20, random_state=0)
```

```
# Counterfactual correctness, all four models side by side
import json, pandas as pd
rows = []
for name, (_, run_dir) in RUNS.items():
    s = json.load(open(f'{run_dir}/summary.json'))
    cc = s['counterfactual_correctness']
    for k, d in cc['contradiction'].items():
        for g in ['all', 'entailment', 'neutral', 'contradiction']:
            x = d[g]
            rows.append(dict(model=name, variant=k, group=g, edits=x['n_edits'], scored=x['n_confident'],
                             should_change=x['oracle_changes_label']['mean'], follows=x['follows_when_should_change']['mean'],
                             stays=x['stays_when_should_stay']['mean']))
    print(name, 'oracle accuracy on original pairs:', round(s['oracle']['accuracy_vs_gold_on_original_pairs'], 3))
cf = pd.DataFrame(rows)
cf[cf.group == 'all'].pivot(index='variant', columns='model', values='follows').round(3)
```

## 5c. Explanation-evidence alignment (all four models)
Run **cells 1 and 2 first**, then this cell. It does NOT need cells 4–5b again.

For 2,000 pairs (the same pairs for every model) the model **writes a new explanation** for each edited input:
premise removal / isolation / contradiction and hypothesis contradiction, cited and control. We then check whether the new
explanation mentions the new word (pool → ocean: does it talk about the ocean?), whether it still cites a word that is
gone, and whether it argues for the label the model now predicts.

Everything else is re-used from the cache, so only the new explanations are computed (~15k generations per model; the
progress line prints the speed and time left). A disconnect is fine: re-run this cell and it continues where it stopped.
The cell also rewrites `summary.json`, which now includes the same-edit-type (op-matched) contradiction contrasts.
At the end it zips the small result files to `btp_results/results_alignment.zip` for download.

```
%cd /content/BTech-Project
!git pull
import os, json, zipfile
RUNS = {
    'random_nt8':       ('/content/drive/MyDrive/nt8',             f'{RESULTS}/t5_nt8_test_full'),
    'fastvotek_nt8':    ('/content/drive/MyDrive/fastvotek_nt8',   f'{RESULTS}/t5_esnli_fastvotek_nt8_full'),
    'random_nt128':     ('/content/drive/MyDrive/random_nt128',    f'{RESULTS}/t5_esnli_random_nt128_full'),
    'fastvotek_nt128':  ('/content/drive/MyDrive/fastvotek_nt128', f'{RESULTS}/t5_esnli_fastvotek_nt128_full'),
}
for name, (ckpt, run_dir) in RUNS.items():
    print(f'\n=================== {name}')
    !python faithfulness/run_experiment.py --model_path "$ckpt" --out_dir "$run_dir" --explain_variants 2000

# small result files only (no records.jsonl / variant_preds.jsonl caches)
out = f'{RESULTS}/results_alignment.zip'
with zipfile.ZipFile(out, 'w', zipfile.ZIP_DEFLATED) as z:
    for name, (_, run_dir) in RUNS.items():
        for f in ['summary.json', 'variant_explanations.csv', 'per_example.csv', 'contradiction_edits.csv']:
            fp = os.path.join(run_dir, f)
            if os.path.exists(fp):
                z.write(fp, os.path.join(os.path.basename(run_dir), f))
        ok = 'explanation_alignment' in json.load(open(os.path.join(run_dir, 'summary.json')))
        print(f'{name:16s} alignment in summary: {ok}')
print(f'Wrote {out}: {os.path.getsize(out)/1e6:.1f} MB  -> download it from Drive')
```
